```python
import pyclf
from pyclf.discriminant_analysis import BinaryLinearDiscriminantAnalysis
from pyclf.datasets import generate_2d_multiclass_data
from pyclf.viz import plot_2d_scatter
from sklearn.metrics import accuracy_score

mus = [
    [-1.0, -1.0],
    [1.0, 1.0],
]

covs = [
    0.1,
    0.1,
]

n_train = [500, 500]
n_test  = [100,  100]

X_train, y_train, X_test, y_test = generate_2d_multiclass_data(
    mus, covs, n_train, n_test, random_state=42
)

#plot_train_test_scatter(X_train, y_train, X_test, y_test)

model = BinaryLinearDiscriminantAnalysis()

model.fit(X_train, y_train)

d = model.decision_function(X_test)
p = model.predict_proba(X_test)
preds = model.predict(X_test)

acc = accuracy_score(y_test, preds)
print(acc)
```

```python
import pyclf
from pyclf.discriminant_analysis import BinaryLinearDiscriminantAnalysis, LinearDiscriminantAnalysis
from pyclf.datasets import generate_2d_multiclass_data
from pyclf.viz import plot_train_test_scatter
from sklearn.metrics import accuracy_score

mus = [
    [0.0, 3.0],
    [-3, -3],
    [3, -3],
]

covs = [
    0.3,
    0.3,
    0.3,
]

n_train = [500, 500, 500]
n_test  = [100,  100,  100]

X_train, y_train, X_test, y_test = generate_2d_multiclass_data(
    mus, covs, n_train, n_test, random_state=42
)

model = LinearDiscriminantAnalysis()
model.fit(X_train, y_train)

model.predict_proba(X_test)
model.predict_log_proba(X_test)
preds = model.predict(X_test)

accuracy_score(y_test, preds)
```
